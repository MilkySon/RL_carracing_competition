# E0 — Original continuous A2C

Professor baseline lineage; outputs separated by experiment and seed. Best/worst videos selected by raw return on seeds 1000–1009.

```python
REPO_URL = "https://github.com/MilkySon/RL_carracing_competition.git"
BRANCH = "main"  # Change to branch containing experiments/continuous_ablation
TEAM = "Reinforcement_material"
SEED = 0
BUDGET = 300_000
EVAL_EVERY = 50_000
EXPERIMENT = "E0"
USE_GOOGLE_DRIVE = True
```

```python
from pathlib import Path
import os, subprocess, sys
if USE_GOOGLE_DRIVE:
    from google.colab import drive
    drive.mount('/content/drive')
    OUTPUT_ROOT = f'/content/drive/MyDrive/CarRacing_A2C_Legacy/outputs'
else:
    OUTPUT_ROOT = '/content/carracing_outputs'
repo_dir = Path('/content/RL_carracing_competition')
if not repo_dir.exists():
    subprocess.run(['git','clone','--branch',BRANCH,REPO_URL,str(repo_dir)],check=True)
base = repo_dir/'experiments'/'continuous_ablation'
assert (base/'scripts'/'run_experiment.py').exists(), 'Commit this package to GitHub branch first.'
subprocess.run([sys.executable,'-m','pip','install','-q','-r',str(base/'baseline'/'requirements.txt')],check=True)
print('Project:',base,'Output:',OUTPUT_ROOT)
```

```python
subprocess.run([sys.executable,str(base/'scripts'/'run_experiment.py'),
    '--experiment',EXPERIMENT,'--team',TEAM,'--seed',str(SEED),
    '--budget',str(BUDGET),'--eval-every',str(EVAL_EVERY),
    '--root',OUTPUT_ROOT],check=True)
```

```python
from IPython.display import display, Video
run_dir=Path(OUTPUT_ROOT)/EXPERIMENT/f'seed_{SEED}'
for label in ('worst','best'):
    videos=sorted((run_dir/'videos').glob(f'{label}_seed_*/*.mp4'))
    print(label, videos)
    if videos: display(Video(str(videos[-1]),embed=True,width=640))
print('ONNX files:', sorted((Path(OUTPUT_ROOT)/'submissions').glob(f'{TEAM}_{EXPERIMENT}_seed{SEED}*.onnx')))
print((run_dir/'reports'/'eval_summary.json').read_text())
```
